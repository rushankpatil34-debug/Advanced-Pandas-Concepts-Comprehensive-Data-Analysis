```python
import pandas as pd

students = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Amit", "Priya", "Neha", "Arjun"],
    "Age": [21, 22, 20, 21, 23],
    "City": ["Mumbai", "Pune", "Mumbai", "Delhi", "Pune"],
    "Marks": [85, 72, 92, 64, 76]
}

df = pd.DataFrame(students)

print(df)
```

```python
#adding coloum
df["Gender"] ="male","male","female","female","male"
print(df)
```

```python
df["Persentage"]=(df["Marks"]*100)/100
print(df)
```

```python
import numpy as np

condition = [
    df["Marks"] >= 90,
    df["Marks"] >= 80,
    df["Marks"] >= 70,
    df["Marks"] >= 60 
]
choice=["A+","A","B","C"]
df["Gread"]=(np.select(condition,choice,default="F"))
print(df)
```

```python
#loc
df["Result"] = "Pass"
df.loc[df["Marks"] < 60,"Result"] = "Fail"
```

```python
print(df)
```

```python
#Adding a new row to the DataFrame using the insert() method
df.insert(2, "Deparment", ["IT", "IT", "CS", "CS", "IT"])
print(df)
```
